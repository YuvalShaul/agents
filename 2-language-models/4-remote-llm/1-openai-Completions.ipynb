## OpenAI Completions lab
- connect to [openai platform](https://openai.com/api/)
- login to the api platform (top right)
- Click on **getting started**
- Choose OS and click on **Create an API key**
- Click on **Create new secret key**
- Make sure you copy is - then save it like this:  
- Open the environment vars (Windows+"env") and save a new key with the name OPENAI_API_KEY and the secret you got as value
- Close and open your VScode

Then try the following:

```python
from openai import OpenAI

# Initialize the client
# client = OpenAI(api_key="your_api_key_here")
client = OpenAI()

# A single interaction (The "Stateless" way)
response = client.chat.completions.create(
    model="gpt-4o",  # or "gpt-3.5-turbo"
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Explain how a photon behaves like both a particle and a wave."}
    ],
    # SECURITY TIP: 
    # Setting store to False (if available in your tier) tells OpenAI 
    # not to persist this specific message in their 'History' dashboard.
    store=False, 
    temperature=0.7
)

# Printing the result
print(response.choices[0].message.content)
```
